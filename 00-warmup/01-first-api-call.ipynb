```python
import requests
import os
from dotenv import load_dotenv
```

```python
load_dotenv("../.env")

CHAT_URL = os.getenv("CHAT_URL")
API_KEY = os.getenv("API_KEY")
MODEL = os.getenv("MODEL")
```

```python
payload = {
    "model": MODEL,
    "max_tokens": 3000,
    "stream": False,
    "messages": [
        {"role": "system", "content": "You are a concise, no-nonsense career advisor."},
        {"role": "user", "content": "In 3 bullet points, what should a software engineer focus on when switching into AI agent engineering?"},
    ],
}
```

```python
response = requests.post(
    CHAT_URL,
    headers={"Authorization": f"Bearer {API_KEY}", "Content-Type": "application/json"},
    json=payload,
)
print("status code:", response.status_code)
```

```python
data = response.json()

print(data["choices"][0]["message"]["content"])
print("\n--- token usage ---")
print(data.get("usage"))
```
